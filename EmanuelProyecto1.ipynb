```python

base = float(input("Ingrese valor de base del rectángulo: "))
altura = float(input("Ingrese valor de altura del rectángulo: "))

área = base*altura
perímetro = 2*(base+altura)

print("El área del rectángulo es:", área)
print("El perímetro del rectángulo es:", perímetro)
```

```python

precio_original = float(input("Ingrese el precio original del producto: "))
descuento = precio_original*15/100
precio_final = precio_original-descuento

print("El precio final del producto es:", precio_final)
```

```python


horas = float(input("Ingrese la cantidad de horas: "))

minutos = horas*60
segundos = horas*3600
print("El equivalente en minutos de las horas ingresadas es:", minutos)
print("El equivalente en segundos de las horas ingresadas es:", segundos)
```
