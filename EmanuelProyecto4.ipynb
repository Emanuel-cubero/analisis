```python
import folium

from folium.plugins import MiniMap

latitud_heredia= 10.000024
longitud_heredia= -84.1165

latitud_cartago= 9.8644
longitud_cartago= -83.9194

latitud_sanjose= 9.9281
longitud_sanjose= -84.0907


my_map = folium.Map(location= [latitud_sanjose, longitud_sanjose], zoom_start=10,
                    titles= "OpenStreeMap")

popuptext_heredia= "Heredia, Costa Rica"
popuptext_cartago= "Cartago, Costa Rica"
popuptext_sanjose= "San Jose, Costa Rica"

folium.Marker(
    [latitud_heredia, longitud_heredia],
    popup=popuptext_heredia,
    tooltip= "Heredia",
    icon=folium.Icon(
        color= "orange",
        icon= "info-sign",
        prefix= "fa")
    ).add_to(my_map)

folium.Marker(
    [latitud_cartago, longitud_cartago],
    popup=popuptext_cartago,
    tooltip= "Cartago",
    icon=folium.Icon(
        color= "blue",
        icon= "info-sign",
        prefix= "fa")
    ).add_to(my_map)

folium.Marker(
    [latitud_sanjose, longitud_sanjose],
    popup=popuptext_sanjose,
    tooltip= "San Jose",
    icon=folium.Icon(
        color= "green",
        icon= "info-sign",
        prefix= "fa")
    ).add_to(my_map)

folium.Circle(
    radius=1000,
    location= [latitud_heredia, longitud_heredia],
    popup=popuptext_heredia,
    color= "black",
    fill=True,
    fill_color= "orange",
    fill_opacity= 0.3,
    tooltip= "Zona Heredia"
).add_to(my_map)

folium.Circle(
    radius=1000,
    location= [latitud_cartago, longitud_cartago],
    popup=popuptext_cartago,
    color= "black",
    fill=True,
    fill_color= "blue",
    fill_opacity= 0.3,
    tooltip= "Zona Cartago"
).add_to(my_map)

folium.Circle(
    radius=10000,
    location= [latitud_sanjose, longitud_sanjose],
    popup=popuptext_sanjose,
    color= "black",
    fill=True,
    fill_color= "green",
    fill_opacity= 0.3,
    tooltip= "Zona San Jose"
).add_to(my_map)

minimap=MiniMap(
    title_layer= "OpenStreetMap",
    position= "bottomright")
my_map.add_child(minimap)

my_map.save("Provincias.html")

my_map
```
