```python
#ejercicio 4
import matplotlib.pyplot as plt

dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]
temperaturas = [18, 22, 31, 27, 16]

for i in range(len(dias)):
    temp = temperaturas[i]
    if temp >= 30:
        clasificacion = "Caluroso"
    elif temp >= 20:
        clasificacion = "Templado"
    else:
        clasificacion = "Frío"
    print(f"{dias[i]}: {temp}°C - {clasificacion}")

plt.bar(dias, temperaturas)
plt.title("Temperaturas durante la semana")
plt.xlabel("Días")
plt.ylabel("Temperatura (°C)")
plt.show()

from google.colab import auth
auth.authenticate_user()
print('Authenticated')
```

```python
#ejercicio 5
import matplotlib.pyplot as plt

calificaciones = [85, 90, 75, 70, 65, 50, 40]

aprobados = 0
reprobados = 0

for nota in calificaciones:
    if nota >= 70:
        aprobados += 1
    else:
        reprobados += 1

print(f"Aprobados: {aprobados}")
print(f"Reprobados: {reprobados}")

categorias = ["Aprobados", "Reprobados"]
cantidades = [aprobados, reprobados]

plt.bar(categorias, cantidades)
plt.title("Resultados de los estudiantes")
plt.ylabel("Cantidad de estudiantes")
plt.show()
```

```python
#ejercicio 6
import matplotlib.pyplot as plt

num = int(input("Ingrese un número para la tabla de multiplicar: "))

multiplicadores = list(range(1, 13, 1))
resultados = []

for i in multiplicadores:
    resultados.append(num * i)

plt.plot(multiplicadores, resultados, marker='o')
plt.title("Tabla de multiplicar")
plt.xlabel("Multiplicador")
plt.ylabel("Resultado")

plt.show()
```

```python
#ejercicio 7

import matplotlib.pyplot as plt

estudiantes = ["Estudiante 1", "Estudiante 2", "Estudiante 3", "Estudiante 4", "Estudiante 5"]
notas = [95, 82, 65, 70, 91]

for i in range(len(notas)):
    if notas[i] >= 90:
        clasificacion = "Excelente"
    elif notas[i] >= 70:
        clasificacion = "Aprobado"
    else:
        clasificacion = "Reprobado"
    print(f"{estudiantes[i]}: {notas[i]} - {clasificacion}")

plt.bar(estudiantes, notas)
plt.title("Calificaciones de los estudiantes")
plt.xlabel("Estudiantes")
plt.ylabel("Calificación")
plt.show()
```

```python
#ejercicio 8
import matplotlib.pyplot as plt

dias = ["Lun", "Mar", "Mié", "Jue", "Vie", "Sáb", "Dom"]
temperaturas = [18, 25, 32, 28, 16, 30, 22]

for i in range(len(dias)):
    temp = temperaturas[i]
    if temp >= 30:
        clasificacion = "Calurosa"
    elif temp >= 20:
        clasificacion = "Agradable"
    else:
        clasificacion = "Fría"
    print(f"{dias[i]}: {temp}°C - {clasificacion}")

plt.plot(dias, temperaturas, marker='o')
plt.title("Temperaturas durante la semana")
plt.xlabel("Días")
plt.ylabel("Temperatura (°C)")
plt.show()
```

```python
#ejercicio 9

import matplotlib.pyplot as plt

dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]
ventas = [120000, 45000, 80000, 150000, 30000]

for i in range(len(dias)):
    venta = ventas[i]
    if venta >= 100000:
        clasificacion = "Venta alta"
    elif venta >= 50000:
        clasificacion = "Venta media"
    else:
        clasificacion = "Venta baja"
    print(f"{dias[i]}: ¢{venta} - {clasificacion}")

plt.bar(dias, ventas)
plt.title("Ventas de la tienda")
plt.xlabel("Días")
plt.ylabel("Ventas en colones")
plt.show()
```

```python
#ejercicio 10

import matplotlib.pyplot as plt

numeros = [5, -3, 0, 12, -7, 0, 8, -1, 4, 0]

positivos = 0
negativos = 0
ceros = 0

for num in numeros:
    if num > 0:
        positivos += 1
    elif num < 0:
        negativos += 1
    else:
        ceros += 1

print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Ceros: {ceros}")

categorias = ["Positivos", "Negativos", "Ceros"]
cantidades = [positivos, negativos, ceros]

plt.bar(categorias, cantidades)
plt.title("Clasificación de números")
plt.xlabel("Categorías")
plt.ylabel("Cantidad")
plt.show()
```
